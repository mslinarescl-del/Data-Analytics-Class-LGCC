```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
url = 'https://raw.githubusercontent.com/jyotibalkar3120-cyber/311_Service_Requests_from_2019May.csv/refs/heads/main/311_Service_Requests_from_2019May.csv'

df = pd.read_csv(url)

print(df.describe())
```

```python
df.columns
```

```python
# How many rows and colums are in this dataset

number = df.shape

print(f"The number of Rows and Columns is {number}.")

```

```python
#What is the time range of this dataset


start = df['Created Date'].min()
end = df['Created Date'].max()

print(f"The dataset covers the period from {start} to {end}.")
```

```python
# What ares some relevant summary statistics

print(df.describe())
```

```python
# create 2-3 visualizations from this dataset

top_complaints = df['Complaint Type'].value_counts().head(10)

top_complaints.plot(kind='barh')

plt.title('Top 10 NYC 311 Complaint Types')
plt.xlabel('Number of Complaints')
plt.ylabel('Complaint Type')
```

```python
borough_counts = df['Borough'].value_counts()

borough_counts.plot(kind='bar')

plt.title('NYC 311 Service Requests by Borough')
plt.xlabel('Borough')
plt.ylabel('Number of Requests')

```

```python
noise_complaints = df[df['Complaint Type'] == 'Noise']

noise_by_borough = noise_complaints['Borough'].value_counts()

noise_by_borough.plot(kind='bar')

plt.title('Noise Complaints by Borough')
plt.xlabel('Borough')
plt.ylabel('Number of Noise Complaints')
```

```python
# What borougth has the most complaints

borough_counts = df['Borough'].value_counts()

print("Borough with the most complaints:", borough_counts.idxmax())
print("Number of complaints is:", borough_counts.max())

```
