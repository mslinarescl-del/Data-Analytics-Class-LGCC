```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
url = 'https://raw.githubusercontent.com/jyotibalkar3120-cyber/311_Service_Requests_from_2019May.csv/refs/heads/main/311_Service_Requests_from_2019May.csv'

df = pd.read_csv(url)

print(df.describe())
```

```python
# How many rows and colums are in this dataset

number = df.shape

print(f"The number of Rows and Columns is {number}.")

```

```python
#What is the time range of this dataset


start = df['Created Date'].min()
end = df['Created Date'].max()

print(f"The dataset covers the period from {start} to {end}.")
```

```python
# What ares some relevant summary statistics

print(df.describe())
```

```python
# create 2-3 visualizations from this dataset

df.hist()
df.plot(kind='scatter', x='Longitude', y='Latitude', title='Geographic Distribution of Complaints')
```

```python
# What borougth has the most complaints

borough_counts = df['Borough'].value_counts()

print("Borough with the most complaints:", borough_counts.idxmax())
print("Number of complaints:", borough_counts.max())

```

```python
df['Borough'].value_counts()
```

```python
# Create a plot of counts by borough for noise complaints

noise_complaints = df[df['Complaint Type'] == 'Noise']

borough_counts = noise_complaints['Borough'].value_counts()

borough_counts.plot(kind='pie')
```
