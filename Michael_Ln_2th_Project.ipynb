```python
import numpy as np
import pandas as pd
from numpy.random  import randn
np.random.seed(123)
```

```python
url = ('https://raw.githubusercontent.com/CunyLaguardiaDataAnalytics/datasets/refs/heads/master/mtcars.csv')
```

```python
df = pd.read_csv(url)
df.head()
```

```python
## understand the data first (head, describe, etc.)

df.describe()
```

```python
df.info()
```

```python
df.columns
```

```python
# clean the dataset by naming the missing column name

df = df.rename(columns={'Unnamed: 0': 'Vehicle'})

df.columns
```

```python
## create a new dataframe that consists of a subset of the original columns (only mpg, hp, wt, and cyl)

df[['mpg','hp','wt','cyl']]
```

```python
# rename the above columns (cylinders = cyl)

df = df.rename(columns={'cyl': 'cylinders'})

df.columns
```

```python
## summary statistics for those columns (average mpg)

df['mpg'].mean()
```

```python
## filter your data and find summary statistics for only one type of vehicle

honda_civic = df[df['Vehicle'] == 'Honda Civic']
honda_civic.describe()
```
